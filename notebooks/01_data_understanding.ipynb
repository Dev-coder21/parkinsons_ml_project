```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

print("Libraries imported successfully!")
```

```python
# Load UCI Parkinson's dataset

parkinsons = fetch_ucirepo(id=174)

X = parkinsons.data.features.copy()
y = parkinsons.data.targets.copy()

print("Dataset loaded successfully!")
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print("Feature columns:")
print(X.columns.tolist())

print("\nData types:")
print(X.dtypes)

print("\nMissing values:")
print(X.isnull().sum())

print("\nTarget distribution:")
print(y.value_counts())

print("\nDuplicate rows:")
print(X.duplicated().sum())
```

```python
import sys
from pathlib import Path

# Allow the notebook to import modules from src/
sys.path.append(str(Path.cwd().parent / "src"))

from preprocessing import prepare_data

X_clean, y_clean = prepare_data(X, y)

print("Clean features shape:", X_clean.shape)
print("Target shape:", y_clean.shape)

print("\nDuplicate feature names:")
print(X_clean.columns[X_clean.columns.duplicated()].tolist())

print("\nFeature names:")
print(X_clean.columns.tolist())
```

```python
# Save the cleaned dataset locally

clean_df = X_clean.copy()
clean_df["status"] = y_clean

clean_df.to_csv(
    "../data/parkinsons.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
print("File: data/parkinsons.csv")
print("Shape:", clean_df.shape)
```
